# 06_v2 rolling evaluation — archived-output verification

This notebook verifies the rolling results reported in the manuscript from the archived output CSV files. **It is not the original fitting notebook.** Locate or reconstruct the original `06_v2_rolling_riskmap_evaluation_previous_week_weather_final` before claiming complete training-code availability.


```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
OUT = ROOT / "reported_outputs"
if not OUT.exists():
    OUT = ROOT.parent.parent / "reported_outputs"

summary = pd.read_csv(OUT / "06_v2_rolling_pooled_summary_previous_week_weather.csv")
events = pd.read_csv(OUT / "06_v2_rolling_event_cases_previous_week_weather.csv")
splits = pd.read_csv(OUT / "06_v2_rolling_split_check_previous_week_weather.csv")
features = pd.read_csv(OUT / "06_v2_rolling_feature_sets_previous_week_weather.csv")
display(summary)
display(splits)
display(features)
```

```python
expected = {
    "geo_season_previous_week_weather": (61, 0.2786885245901639, 0.45901639344262296),
    "geo_season_previous_week_weather_same_grid_lags": (61, 0.26229508196721313, 0.45901639344262296),
    "geo_season_previous_week_weather_neighbor_past": (61, 0.26229508196721313, 0.45901639344262296),
    "geo_season_previous_week_weather_lags_neighbor_past": (61, 0.21311475409836064, 0.4426229508196721),
}
for model, (n, top10, top20) in expected.items():
    row = summary.loc[summary.model_name.eq(model)].iloc[0]
    assert int(row.total_events) == n
    assert abs(float(row.top10_capture_rate) - top10) < 1e-12
    assert abs(float(row.top20_capture_rate) - top20) < 1e-12
assert set(events.model_name.unique()) == set(expected)
assert events.groupby("model_name").size().eq(61).all()
print("Archived rolling outputs match the manuscript Table 2 values.")
```
